```python
# TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
```

```python
# TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()

        keywords = list(
            set([
                w.lower()
                for w in words
                if len(w) > 4
            ])
        )

        return keywords[:5]

    except Exception:
        return []
```

```python
#  SINGLE AGENT

def agent(query: str):
    try:
        # Convert query to lowercase
        query_lower = query.lower()


        # Route 1: Calculation

        if "calculate" in query_lower:

            # Remove the routing keyword
            expression = query_lower.replace(
                "calculate", "", 1
            ).strip()

            # Check for missing expression
            if not expression:
                return {
                    "type": "error",
                    "result": "No mathematical expression provided"
                }

            # Call calculator tool
            result = calculator(expression)

            # Handle calculator error
            if result == "Error in calculation":
                return {
                    "type": "error",
                    "result": result
                }

            return {
                "type": "calculation",
                "result": result
            }


        # Route 2: Keyword Extraction

        elif "keywords" in query_lower:

            # Remove the routing keyword
            text = query_lower.replace(
                "keywords", "", 1
            ).strip()

            # Check for missing text
            if not text:
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction"
                }

            # Call keyword extraction tool
            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }


        # Route 3: General Query

        else:
            return {
                "type": "general",
                "result": f"I received your query: {query}"
            }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }
```

```python
# TEST CASES

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
```
